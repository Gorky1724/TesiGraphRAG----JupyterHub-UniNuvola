```python
# Attualmente impostato download dataset1 (su RAG) -- da arXiv
import os
import urllib.request

# Cartella di destinazione
### Cambiare se necessario istanziare altri dataset
target_dir = os.path.expanduser("~/tesi_graphrag/data/raw/ds1")

os.makedirs(target_dir, exist_ok=True)

# I 10 paper selezionati per il dataset
### Cambiare se necessario istanziare altri dataset
papers = {
    "GraphRAG_Microsoft": "https://arxiv.org/pdf/2404.16130.pdf",
    "Self_RAG": "https://arxiv.org/pdf/2310.11511.pdf",
    "CRAG_Corrective_RAG": "https://arxiv.org/pdf/2401.15884.pdf",
    "RAPTOR_Hierarchical_RAG": "https://arxiv.org/pdf/2401.18059.pdf",
    "Original_RAG_Lewis": "https://arxiv.org/pdf/2005.11401.pdf",
    "HyDE_Hypothetical_Embeddings": "https://arxiv.org/pdf/2212.10496.pdf",
    "KG_RAG_Survey": "https://arxiv.org/pdf/2402.11199.pdf",
    "FLARE_Active_RAG": "https://arxiv.org/pdf/2305.14283.pdf",
    "Knowledge_Graph_Prompting": "https://arxiv.org/pdf/2308.00352.pdf",
    "RAG_vs_FineTuning": "https://arxiv.org/pdf/2401.08406.pdf",
}

print(f"Inizio download dei {len(papers)} paper in: {target_dir}\n")

for name, url in papers.items():
    file_path = os.path.join(target_dir, f"{name}.pdf")
    if not os.path.exists(file_path):
        print(f"-> Scaricando {name}.pdf...")
        try:
            req = urllib.request.Request(
                url, headers={"User-Agent": "Mozilla/5.0"}
            )
            with (
                urllib.request.urlopen(req) as response,
                open(file_path, "wb") as out_file,
            ):
                out_file.write(response.read())
            size_kb = os.path.getsize(file_path) // 1024
            print(f"   [OK] Completato ({size_kb} KB)")
        except Exception as e:
            print(f"   [ERRORE] Impossibile scaricare {name}: {e}")
    else:
        print(f"-> {name}.pdf già presente.")

print("\n!>> Download completato")
```
