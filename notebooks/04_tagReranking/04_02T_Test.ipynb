```python
import sys
import json
from pathlib import Path

project_root = Path("~/tesi_graphrag").expanduser().resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.graph_builder import KnowledgeGraphBuilder
from src.sidecar_manager import SidecarManager
from src.chunk_widget import ChunkGraphWidget

print("!>> Moduli caricati con successo!")
```

```python
# Inizializzazione del gestore dati sidecar
sidecar_path = project_root / "data/processed/test/tag_test_sidecar.json"
sidecar = SidecarManager(sidecar_path)

# Inizializzazione del gestore dati sidecar (reset per un test pulito)
sidecar.reset_all()

# Istanziazione del Builder del Grafo
builder = KnowledgeGraphBuilder()

# Inserimento Nodi Chunk
builder.add_chunk_node("chunk_0", text="GraphRAG combina Vector DB e Grafi di Conoscenza.", tags=["GraphRAG"])
builder.add_chunk_node("chunk_1", text="Qdrant gestisce il recupero dei vettori ad alte prestazioni.", tags=["VectorDB"])
builder.add_chunk_node("chunk_2", text="NetworkX modella la struttura del grafo nel backend Python.", tags=["NetworkX"])
builder.add_chunk_node("chunk_3", text="D3.js gestisce il rendering interattivo del frontend.", tags=["Frontend"])

# Inserimento Archi/Relazioni di similarità
builder.add_relation("chunk_0", "chunk_1", weight=0.85)
builder.add_relation("chunk_0", "chunk_2", weight=0.78)
builder.add_relation("chunk_0", "chunk_3", weight=0.60)
builder.add_relation("chunk_1", "chunk_2", weight=0.50)

# Esportazione JSON per D3.js
graph_data = builder.to_json_data()

# Sincronizzazione dei tag iniziali nel SidecarManager
for node in graph_data.get("nodes", []):
    for tag in node.get("tags", []):
        sidecar.add_tag_override(node["id"], tag)

print(f"!> Grafo generato: {len(graph_data['nodes'])} nodi e {len(graph_data['links'])} archi.")
print(f"!> Tag iniziali registrati nel Sidecar: {list(sidecar.get_global_tags().keys())}")
```

```python
# Cella 3: Istanziazione e Visualizzazione del Widget

# 1. Inizializzazione del widget con il file sidecar dedicato
widget = ChunkGraphWidget(sidecar_path=sidecar_path)

# 2. Caricamento del grafo (applica in automatico distanze e tag salvati nel sidecar)
widget.load_graph(graph_data)

# 3. Rendering del widget interattivo
widget
```
